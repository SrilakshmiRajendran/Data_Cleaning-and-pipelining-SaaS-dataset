Task 4 – Data Cleaning Pipeline

Create functions that perform the following cleaning steps.

1. Handle Missing Values

Rules:

● If the name is missing → replace it with "Unknown Customer."
● If usage_hours is NULL or missing → replace with 0
● If monthly_fee is invalid → mark it as None

2. Remove Duplicate Records
● Duplicate customers may appear in the dataset.
● Use customer_id to detect duplicates.
● Create a function: remove_duplicates(data), which removes the duplicates.

3. Standardize Date Formats
● Convert signup_date to the standard format: YYYY-MM-DD
● Use the datetime module to parse dates.

```python
print(saas_data)
```

```python
class SaaSCustomer:

    def __init__(self, customer_id, name, signup_date,
                 plan, monthly_fee, usage_hours):

        self.customer_id = customer_id
        self.name = name
        self.signup_date = signup_date
        self.plan = plan
        self.monthly_fee = monthly_fee
        self.usage_hours = usage_hours

    def validate_data(self):

        # Convert monthly_fee to float
        try:
            self.monthly_fee = float(self.monthly_fee)
        except:
            self.monthly_fee = 0.0

        # Convert usage_hours to integer
        try:
            self.usage_hours = int(self.usage_hours)
        except:
            self.usage_hours = 0

        # Replace missing name
        if self.name is None or self.name == "":
            self.name = "Unknown Customer"

        # Standardize plan
        if self.plan == "":
            self.plan = "Unknown"
        else:
            self.plan = self.plan.strip().title()

    def display_customer(self):

        print(
            self.customer_id,
            self.name,
            self.signup_date,
            self.plan,
            self.monthly_fee,
            self.usage_hours
        )


# -----------------------------------
# CREATE OBJECTS AND REMOVE DUPLICATES
# -----------------------------------

customers = []

# Set to remember customer IDs already processed
seen_ids = set()

for i in range(len(saas_data["customer_id"])):

    customer_id = saas_data["customer_id"][i]

    # Check whether customer ID is already present
    if customer_id not in seen_ids:

        customer = SaaSCustomer(
            saas_data["customer_id"][i],
            saas_data["name"][i],
            saas_data["signup_date"][i],
            saas_data["plan"][i],
            saas_data["monthly_fee"][i],
            saas_data["usage_hours"][i]
        )

        # Clean the customer's data
        customer.validate_data()

        # Store object
        customers.append(customer)

        # Remember this customer ID
        seen_ids.add(customer_id)


# -----------------------------------
# DISPLAY CLEANED CUSTOMERS
# -----------------------------------

for customer in customers:
    customer.display_customer()
```

Calculate and print the following:

Note: The following output is calculated after cleaning and deduplication have been applied.



```python
# Total number of customers
total_customers = len(customers)

# Average monthly fee
total_fee = sum(customer.monthly_fee for customer in customers)
average_fee = total_fee / len(customers)

# Total usage hours
total_usage = sum(customer.usage_hours for customer in customers)

# Number of customers per plan
plan_count = {}

for customer in customers:

    plan = customer.plan

    if plan in plan_count:
        plan_count[plan] += 1
    else:
        plan_count[plan] = 1


# Display results
print("Total Customers:", total_customers)
print("Average Monthly Fee:", round(average_fee, 2))
print("Total Usage Hours:", total_usage)

print("\nPlan Distribution")

for plan, count in plan_count.items():
    print(plan, "-", count)
```

Task 6 – Identify Low-Usage Customers

Create a function called find_low_usage_customers().

Rule: Customers with usage_hours less than 50 should be identified.

Example Output

Low Usage Customers ID: 105 | Name: David | Plan: Pro | Usage Hours: 5 ID: 102 | Name: Meena Nair | Plan: Basic | Usage Hours: 10

● Use try, except, else, and finally to handle file errors. ● Also, create another file named plan_summary.txt.

Example Content Pro : 3 Basic: 1 Enterprise: 1 Unknown: 1



```python
## FIND LOW-USAGE CUSTOMERS

def find_low_usage_customers(customers):

    low_usage_customers = []

    for customer in customers:

        if customer.usage_hours < 50:
            low_usage_customers.append(customer)

    return low_usage_customers


low_usage = find_low_usage_customers(customers)


print("\n========== LOW USAGE CUSTOMERS ==========")

for customer in low_usage:

    print(
        "ID:", customer.customer_id,
        "| Name:", customer.name,
        "| Plan:", customer.plan,
        "| Usage Hours:", customer.usage_hours
    )


# =========================================================
# 11. CREATE plan_summary.txt
# =========================================================

file = None

try:

    file = open("plan_summary.txt", "w")

    for plan, count in plan_count.items():
        file.write(f"{plan} : {count}\n")

except Exception as e:

    print("Error while writing file:", e)

else:

    print("\nplan_summary.txt created successfully.")

finally:

    if file is not None:
        file.close()
```
